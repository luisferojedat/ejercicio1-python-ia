### A.2 — Contar vocales (10 pts)

Crea una función `contar_vocales(texto)` que reciba un string y devuelva un diccionario con el conteo de cada vocal (a, e, i, o, u). Debe ser case-insensitive (mayúsculas = minúsculas). No contar acentos.

```python
def contar_vocales(texto: str) -> dict:
    vocales = {'a': 0, 'e': 0, 'i': 0, 'o': 0, 'u': 0}
    texto = texto.lower()  # Convertir a minúsculas para que no distinga entre mayúsculas y minúsculas
    for char in texto:
        if char in vocales:
            vocales[char] += 1
    return vocales
```

```python
# Validación A.2 — NO MODIFICAR
assert contar_vocales("Hola Mundo") == {"a": 1, "e": 0, "i": 0, "o": 2, "u": 1}, "Error: Hola Mundo"
assert contar_vocales("") == {"a": 0, "e": 0, "i": 0, "o": 0, "u": 0}, "Error: vacío"
assert contar_vocales("AEIOU") == {"a": 1, "e": 1, "i": 1, "o": 1, "u": 1}, "Error: AEIOU"
assert contar_vocales("xyz") == {"a": 0, "e": 0, "i": 0, "o": 0, "u": 0}, "Error: sin vocales"
print("✅ A.2 — contar_vocales: CORRECTO")
```

```python
#DESARROLLO LUISF
def contar_vocales(texto: str) -> dict:
    vocales = {'a': 0, 'e': 0, 'i': 0, 'o': 0, 'u': 0}
    texto = texto.lower()  # Ignorar mayúsculas

    for char in texto:
        if char in vocales:
            vocales[char] += 1

    return vocales


# Validación A.2 — NO MODIFICAR
assert contar_vocales("Hola Mundo") == {"a": 1, "e": 0, "i": 0, "o": 2, "u": 1}, "Error: Hola Mundo"
assert contar_vocales("") == {"a": 0, "e": 0, "i": 0, "o": 0, "u": 0}, "Error: vacío"
assert contar_vocales("AEIOU") == {"a": 1, "e": 1, "i": 1, "o": 1, "u": 1}, "Error: AEIOU"
assert contar_vocales("xyz") == {"a": 0, "e": 0, "i": 0, "o": 0, "u": 0}, "Error: sin vocales"
print("✅ A.2 — contar_vocales: CORRECTO")


# Uso con input y validación de longitud
max_caracteres = 100 # Define el número máximo de caracteres

while True:
    texto_usuario = input(f"Ingresa un texto (máximo {max_caracteres} caracteres): ")
    if len(texto_usuario) <= max_caracteres:
        break
    else:
        print(f"Error: El texto excede el límite de {max_caracteres} caracteres. Inténtalo de nuevo.")

resultado = contar_vocales(texto_usuario)

print("Conteo de vocales:")
print(resultado)
```

```python
#DESARROLO LUISFER
def aplanar(lista: list) -> list:
    resultado = []
    for elemento in lista:
        if isinstance(elemento, list):
            resultado.extend(aplanar(elemento))
        else:
            resultado.append(elemento)
    return resultado


# Entrada del usuario (ejemplo: [1, [2, [3, 4]]])
entrada = input("Ingresa una lista: ")

# Convertir texto a lista
lista = eval(entrada)

# Mostrar resultado
print("Lista aplanada:", aplanar(lista))
```
